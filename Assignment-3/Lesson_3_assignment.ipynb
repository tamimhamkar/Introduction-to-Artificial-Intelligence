Data Types and Variables

Section 2: Lists

```python
# 2. Lists

books = ["Python Basics", "Advanced Python", "Data Science with Python"]

# Add two more books to the list
books.append("Machine Learning Basics")
books.append("Artificial Intelligence")

print("Updated book list:", books)
```

Section 3: Loops

```python
# 3. Loops

# Print each book with its index
for i in range(len(books)):
    print(i, books[i])
```

Section 4: User Input

```python
# 4. User Input

new_book = input("Enter a new book title: ")
books.append(new_book)

print("Updated book list:", books)
```

Section 5: Functions

```python
# 5. Functions

# Check if a book is available in the library
def check_availability(book_name):
    if book_name in books:
        return True
    else:
        return False

# Test the function
print("Python Basics available:", check_availability("Python Basics"))
print("Java Programming available:", check_availability("Java Programming"))
```

Section 6: File Handling

```python
# 6. File Handling - Write books to a file

try:
    with open("books.txt", "w") as file:
        for book in books:
            file.write(book + "\n")

    print("Books saved to books.txt successfully.")

except Exception as e:
    print("Error saving books:", e)
```

Read books from the file

```python
# File Handling - Read books from a file

def read_books_from_file(filename):
    try:
        with open(filename, "r") as file:
            books_from_file = file.read().splitlines()
        return books_from_file

    except FileNotFoundError:
        print("Error: File not found.")
        return []

# Test the function
loaded_books = read_books_from_file("books.txt")

print("Books read from file:")
for book in loaded_books:
    print(book)
```

Section 7: Pickle

```python
# 7. Pickle - Save books

try:
    with open("books.pkl", "wb") as file:
        pickle.dump(books, file)

    print("Books saved with Pickle successfully.")

except Exception as e:
    print("Error saving Pickle file:", e)
```

Load books using Pickle

```python
# Pickle - Load books

def load_books_pickle(filename):
    try:
        with open(filename, "rb") as file:
            loaded_books = pickle.load(file)
        return loaded_books

    except FileNotFoundError:
        print("Error: Pickle file not found.")
        return []

# Test the function
pickle_books = load_books_pickle("books.pkl")

print("Books loaded from Pickle:")
for book in pickle_books:
    print(book)
```

Section 8: SQLite Database

```python
# 8. SQLite Database - Create database and table

try:
    conn = sqlite3.connect("library.db")
    cursor = conn.cursor()

    cursor.execute("""
    CREATE TABLE IF NOT EXISTS books (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT NOT NULL
    )
    """)

    conn.commit()
    conn.close()

    print("Database and books table created successfully.")

except sqlite3.Error as e:
    print("Database error:", e)
```

Add a book to the SQLite database

```python
# SQLite Database - Add a book

def add_book_to_db(title):
    try:
        conn = sqlite3.connect("library.db")
        cursor = conn.cursor()

        cursor.execute(
            "INSERT INTO books (title) VALUES (?)",
            (title,)
        )

        conn.commit()
        conn.close()

        print(f"Book '{title}' added to database successfully.")

    except sqlite3.Error as e:
        print("Database error:", e)

# Test the function
add_book_to_db("Python Basics")
```
