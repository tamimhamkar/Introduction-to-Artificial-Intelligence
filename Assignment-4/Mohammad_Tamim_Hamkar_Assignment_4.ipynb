```python
# Import required libraries
import pandas as pd
import numpy as np
from functools import reduce
```

Task 1 — Create a DataFrame

```python
# Task 1: Create a DataFrame

# Create the DataFrame with Name, Age, and City
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)

# Display the DataFrame and check the data types
print(df)
print("\nData Types:")
print(df.dtypes)
```

Task 3 — Handling Null Values

```python
# Task 3: Handling Null Values

# Create a DataFrame with null values
df_null = pd.DataFrame({
    'Name': ['Alice', None, 'Charlie', 'David'],
    'Age': [25, 30, np.nan, 28],
    'City': ['New York', 'Toronto', 'Chicago', None]
})

print("DataFrame with null values:")
print(df_null)

# Fill null values with appropriate values
df_filled = df_null.copy()
df_filled['Name'] = df_filled['Name'].fillna('Unknown')
df_filled['Age'] = df_filled['Age'].fillna(0)
df_filled['City'] = df_filled['City'].fillna('Unknown')

print("\nDataFrame after filling null values:")
print(df_filled)

# Another method is to remove rows containing null values
print("\nDataFrame after using dropna():")
print(df_null.dropna())
```

Task 4 — GroupBy and Describe

```python
# Task 4: GroupBy and Describe

# Create the DataFrame
df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# Group the data by Category and describe the Value column
group_stats = df_group.groupby('Category')['Value'].describe()

# Display the statistics for each category
print(group_stats)

# Interpret the results
print("\nInterpretation:")
print("Category A has 3 values with an average of 18.33.")
print("Category B has 2 values with an average of 22.50.")
print("Category C has 1 value with an average of 35.00.")
```

Task 5: Concatenation and Merging

```python
# Task 5: Concatenation and Merging

# Create the DataFrames
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Concatenate df1 and df2 vertically
df_combined = pd.concat([df1, df2], ignore_index=True)

print("After vertical concatenation:")
print(df_combined)

# Combine df3 horizontally with the concatenated DataFrame
df_final = pd.concat([df_combined, df3], axis=1)

print("\nAfter horizontal merge:")
print(df_final)

# Display the final shape
print("\nFinal shape:", df_final.shape)
```

Task 6 — Tuples and Sets

```python
# Create a tuple of fruits and a set of numbers
fruits = ('apple', 'banana', 'orange')
numbers = {1, 2, 3, 4, 5}

print("Original tuple:", fruits)
print("Original set:", numbers)

# Try to add an item to the tuple
try:
    fruits.add('mango')
except AttributeError:
    print("\nCannot add an item to a tuple because tuples are immutable.")

# Add an item to the set
numbers.add(6)
print("Set after adding 6:", numbers)

# Explain the difference
print("\nTuples cannot be changed after creation, but sets can be changed.")
```

Task 8 — Functions and Lambda

```python
# Task 8: Functions and Lambda

# Regular function to calculate the square of a number
def square(number):
    return number ** 2

# Lambda function to calculate the square
square_lambda = lambda number: number ** 2

# Test both functions with two different numbers
print("Regular function:")
print("Square of 4:", square(4))
print("Square of 7:", square(7))

print("\nLambda function:")
print("Square of 4:", square_lambda(4))
print("Square of 7:", square_lambda(7))
```

Task 9 — Iterators and Generators

```python
# Task 9: Iterators and Generators

# Iterator class for the first 5 even numbers
class EvenIterator:
    def __init__(self):
        self.number = 2
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 5:
            current = self.number
            self.number += 2
            self.count += 1
            return current
        else:
            raise StopIteration

# Generator function for the first 5 even numbers
def even_generator():
    for number in range(2, 11, 2):
        yield number

# Display the results
print("Iterator:", list(EvenIterator()))
print("Generator:", list(even_generator()))
```

Task 10: Map, Reduce, and Filter

```python
# Square all numbers using map()
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Find the product of all numbers using reduce()
product = reduce(lambda x, y: x * y, numbers)

# Get only the even numbers using filter()
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

# Display the results
print("Squared numbers:", squared_numbers)
print("Product of all numbers:", product)
print("Even numbers:", even_numbers)
```

Task 11 — Object-Oriented Programming

```python
# Task 11: Create a Rectangle class

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Calculate the area
    def area(self):
        return self.length * self.width

    # Calculate the perimeter
    def perimeter(self):
        return 2 * (self.length + self.width)

# Create two Rectangle objects
rectangle1 = Rectangle(5, 3)
rectangle2 = Rectangle(8, 4)

# Display the results
print("Rectangle 1:")
print("Area:", rectangle1.area())
print("Perimeter:", rectangle1.perimeter())

print("\nRectangle 2:")
print("Area:", rectangle2.area())
print("Perimeter:", rectangle2.perimeter())
```

Task 12: Pandas Data Analysis

```python
# Task 12: Pandas Data Analysis

# Create the employee DataFrame
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# Calculate the average salary by department
average_salary = df_employees.groupby('Department')['Salary'].mean()

print("Average salary by department:")
print(average_salary)

# Get employees with salary greater than 60000
high_salary_employees = df_employees[df_employees['Salary'] > 60000]['Name']

print("\nEmployees with salary greater than 60000:")
print(high_salary_employees.tolist())

# Add a Bonus column equal to 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.10

print("\nDataFrame with Bonus column:")
print(df_employees)
```
